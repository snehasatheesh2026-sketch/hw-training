```python
jsondf = pd.read_json("users.json")

jsondf
```

```python
json_to_parquet  = jsondf.to_parquet("users.parquet", engine="pyarrow", index= False)

json_to_parquet

p_jdf = pd.read_parquet("users.parquet", engine="pyarrow")

p_jdf
```

```python
# converting parquet to json file
p_jdf.to_json("users_parquet_to_json", orient="records", indent=4)

```

<!--  orient="records"

This controls the structure of the JSON.

Suppose your DataFrame is:

Name	Age
John	25
Sara	30

With:

orient="records"

Output:

[
  {
    "Name": "John",
    "Age": 25
  },
  {
    "Name": "Sara",
    "Age": 30
  }
] -->

<!-- Each row becomes a JSON object (record).

Without orient="records" (default):

{
  "Name":{"0":"John","1":"Sara"},
  "Age":{"0":25,"1":30}
} -->

<!-- indent=4

This makes the JSON readable by adding spaces and line breaks -->
