```python
import requests
from parsel import Selector


class WhiskySpider:

    def __init__(self):
        # URL configuration
        self.start_url = "https://www.whiskyshop.com/scotch-whisky?item_availability=In+Stock"
        self.headers = {
            "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36"
        }
        self.all_items = []

    def parse(self, url):
        print(f"Scraping page: {url}")
        response = requests.get(url, headers=self.headers)

        if response.status_code != 200:
            return

        selector = Selector(text=response.text)

        # Loop through each product card container
        products_list = selector.xpath(
            '//div[contains(@class, "product-item-info")]'
        )
        for products in products_list:
            try:
                item_data = {
                    "name": products.xpath(
                        './/a[contains(@class, "product-item-link")]/text()'
                    )
                    .get()
                    .strip(),
                    "price": products.xpath('.//span[@class="price"]/text()')
                    .get()
                    .replace("£", "")
                    .strip(),
                    "link": products.xpath(
                        './/a[contains(@class, "product-item-link")]/@href'
                    ).get(),
                }
                self.all_items.append(item_data)
            except Exception:
            
                item_data = {
                    "name": products.xpath(
                        './/a[contains(@class, "product-item-link")]/text()'
                    )
                    .get()
                    .strip(),
                    "price": "sold out",
                    "link": products.xpath(
                        './/a[contains(@class, "product-item-link")]/@href'
                    ).get(),
                }
                self.all_items.append(item_data)

        # --- PAGINATION LOGIC
        next_page = selector.xpath(
            '//a[contains(@class, "action") and contains(@class, "next")]/@href'
        ).get()

        if next_page is not None:
            # follow the next page link 
            self.parse(next_page)

    def run(self):
        
        self.parse(self.start_url)

        print(f"\nTotal items collected: {len(self.all_items)}")
        for item in self.all_items:
            print(item)


# To run the script:
spider = WhiskySpider()
spider.run()
```

```python
import pandas as pd

#
df = pd.DataFrame(spider.all_items)

df.to_json("whisky.json", orient="records", indent=4)

df.to_csv("whisky.csv", index= False)
```
