```python
import requests
import csv
from parsel import Selector


class LidlSpider:

    def __init__(self):
        self.start_url = "YOUR_URL"

        self.headers = {
            ""
        }

        self.cookies = {
            "cookie_name": "cookie_value"
        }


    def get_page(self, url):
        try:
            response = requests.get(url)
            response.raise_for_status()
            return Selector(response.text)

        except Exception as e:
            print(f"Error fetching page: {e}")
            return None

    def parse(self, url):

        selector = self.get_page(url)

        if selector is None:
            return

        products = selector.xpath('//YOUR_PRODUCT_XPATH')

        for product in products:

            item = {
                "product_name": product.xpath(
                    'normalize-space(.//strong[@class="product name product-item-name"])'
                ).get(),

                "product_url": product.xpath(
                    './/a/@href'
                ).get(),

                "website": "lidl.ch"
            }

            yield item

        next_page = selector.xpath(
            '//a[contains(@class,"next")]/@href'
        ).get()

        if next_page:
            yield from self.parse(next_page)


spider = LidlSpider()

with open("lidl_products.csv", "w", newline="", encoding="utf-8") as file:

    writer = csv.DictWriter(
        file,
        fieldnames=[
            "product_name",
            "website",
            "product_url"
        ]
    )

    writer.writeheader()

    for item in spider.parse(spider.start_url):
        writer.writerow(item)

print("Data saved successfully")
```
