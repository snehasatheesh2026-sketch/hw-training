````python
# Private variables (using __ double underscore)

class Bank:
    def __init__(self, balance):
        self.__balance = balance   # private variable

    def show_balance(self):
        print(self.__balance)

b1 = Bank(1000)
b1.show_balance()
````

````python
# Inheritance in Python 

# Inheritance means:

# One class (child) can use the properties and methods of another class (parent).
````

````python
class Parent:
    def show(self):
        print("This is Parent class")

class Child(Parent):
    pass

c = Child()
c.show()
````

````python
# Polymorphism
# Same function name can behave differently depending on situation.


# Same interface, different behavior
# ✔ Reduces complexity
# ✔ Improves flexibility
# ✔ Clean code design
````

````python
class Dog:
    def sound(self):
        print("Bark")

class Cat:
    def sound(self):
        print("Meow")

d = Dog()
c = Cat()

d.sound()
c.sound()
````

````python
# Abstraction
# ---------------------------------------

# Abstraction means:

# Hiding implementation details and showing only the important features


# Hides internal details
# ✔ Shows only required things
# ✔ Increases security
# ✔ Reduces complexity
````

````python
from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):
    def sound(self):
        print("Bark")

class Cat(Animal):
    def sound(self):
        print("Meow")

d = Dog()
c = Cat()

d.sound()
c.sound()



# Hidden:
# How sound() is implemented in parent
# Internal design rule
# ✔ Shown:
# Only method name (sound)
# Required structure
````

````python
class Student:
    def __init__(self):
        print("Object created")

    def __del__(self):
        print("Object destroyed")

s1 = Student()
del s1
````

````python
# Difference between Constructor and Destructor
# Feature	Constructor	Destructor
# Method	__init__	__del__
# When runs	Object creation	Object deletion
# Purpose	Initialize data	Cleanup memory
# Call	Automatic




# I think you mean **“usage”** 

# ---

# #  Usage of Constructor and Destructor

# ##  1. Constructor (`__init__`) – Usage

# ###  Used for:

# * Initializing object values
# * Setting up starting data
# * Automatic setup when object is created

# ###  Example use cases:

# * Student details (name, age)
# * Bank account balance setup
# * Product details in e-commerce

# ```python id="u1"
# class Student:
#     def __init__(self, name):
#         self.name = name
# ```

# ---

# ## 2. Destructor (`__del__`) – Usage

# ###  Used for:

# * Cleaning up resources
# * Closing files / connections
# * Freeing memory (rarely needed manually in Python)

# ### Example use cases:

# * Closing database connections
# * Releasing file resources
# * Cleanup before object is removed

# ```python id="u2"
# class Test:
#     def __del__(self):
#         print("Object removed")
# ```

# ---

# # Simple Interview Answer

# > "Constructor is used to initialize object data when it is created. Destructor is used to clean up resources when the object is deleted or destroyed."

# ---

# #  Easy Memory Trick

# * Constructor → setup start 
# * Destructor → cleanup end 

# ---

# If you want  I can next give:
#  Full OOP revision in 1 page (very useful for interviews)
````
