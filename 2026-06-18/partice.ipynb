````python
# Class in Python

# A class is a blueprint (template) for creating objects.

# 👉 It defines properties (variables) and behaviors (functions).

# Example:
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
````

````python
# 2. Object in Python

# An object is an instance of a class.

#  If class is a blueprint, object is the real thing created from it.

# Example:
s1 = Student("Shine", 22)
s2 = Student("John", 25)
````

````python
# . Why OOP?

# Before OOP, people wrote only functions like this:

# name = "Shine"
# age = 22

# def show():
#     print(name, age)

# show()

# 👉 Problem:

# Data and functions are separate
# Not organized for big programs

# Data + Functions together = OBJECT

# SUPER SIMPLE UNDERSTANDING
# Concept	Meaning
# Class	Blueprint 
# Object	Real thing 
# self	Current object 
# def	Function inside class 
````

````python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def show_details(self):
        print("Name:", self.name)
        print("Age:", self.age)

# creating object
s1 = Student("Shine", 22)

# calling method
s1.show_details()
````

````python
2.# Positional arguments(args), Keyword arguments(kwargs)
# ---------------------------------------------------------------

# 1. *args — Usage
# Purpose:

# Used when you don’t know how many positional arguments will be passed.

#  Real usage:
# Sum of numbers
# Calculations with variable inputs
# Passing multiple values without fixing parameter count


# 2. **kwargs — Usage
# 👉 Purpose:

# Used when you don’t know how many named (key-value) arguments will be passed.

# *args → many values (no names)
# **kwargs → many values (with names)

# "*args is used to pass a variable number of positional arguments, and **kwargs is used to pass a variable number of keyword arguments. They make functions flexible and useful when input size is not fixed."

# Where it is used in real life:
# User registration forms (different fields for different users)
# APIs (sending dynamic data)
# Configuration settings (flexible options)
````

````python
# Positional arguments(args)
# -------------------------------------
# args allows a function to accept multiple positional arguments.
# The arguments are stored as a tuple."

def add(*args):
    return sum(args)

add (1,2,3,4)
````

````python
# Keyword Arguments (**kwargs)
# -------------------------------------------

# **kwargs allows a function to accept multiple keyword arguments.
# The arguments are stored as a dictionary."

def display(**kwargs):
    return kwargs

display(name="Shine", age=22)
````

````python
# 4. Generators

# A generator is a special type of function that uses the yield keyword instead of return.

# It generates values one at a time, which saves memory.

# Why Use Generators?
# 1. Memory efficient

# They don’t store all values in memory.

# 2. Faster for large data

# Useful when data is huge.

# 3. Lazy execution

# Values are generated only when needed.

# Real-Life Usage
# Reading large files line by line
# Big data processing
# Streaming data (YouTube, logs, sensors)

# Why yield is important?

# If you used return instead:

# def my_function():
#     return [1, 2, 3, 4, 5]

# 👉 This stores ALL values in memory at once

# But yield:

# gives one value
# pauses function
# continues later
````

````python
def my_generator():
    for i in range(1, 6):
        yield i

````

````python
5# retrun vs yield
# ---------------------------------
````

````python
# 1. return
# Meaning:

# return sends a final result and stops the function.

# Example:
# def demo():
#     return 10

# print(demo())
````

````python
# yield
# Meaning:

# yield gives values one by one and pauses the function.

# Example:
# def demo():
#     yield 1
#     yield 2
#     yield 3

# for value in demo():
#     print(value)

# Output:

# 1
# 2
# 3
# Key points:

def count_up_to(max_val):
    count = 1
    while count <= max_val:
        yield count
        count += 1

# Instantiate the generator
counter = count_up_to(3)

# Retrieve values using next()
print(next(counter))  # Outputs: 1
print(next(counter))  # Outputs: 2
print(next(counter))  # Outputs: 3
# Calling next(counter) again would raise a StopIteration exception

# Function does NOT end immediately
# Pauses and resumes
# Produces multiple values


# Main Difference
# Feature	return	yield
# Output	One value	Multiple values
# Execution	Stops function	Pauses function
# Memory	Stores result	Generates one by one
# Type	Normal function	Generator function
````

````python
6# try and except in Python (Exception Handling)
# ------------------------------------------------------------

# try and except are used to handle errors safely without stopping the program.


# 1. try block
# Meaning:

# Code that might cause an error is written inside try.

# 2. except block
# Meaning:

# If an error happens, Python runs the code inside except
````

````python
try:
    x = 10 / 0
    print(x)
except:
    print("Error occurred")
````

````python
def divide(a, b):
    try:
        result = a / b
        return result
    except ZeroDivisionError:
        return "Cannot divide by zero"

print(divide(10, 2))
print(divide(10, 0))
````

````python
# Why we use try-except inside functions?
# To prevent function crash
# To handle errors safely inside reusable code
# To return clean output instead of error
````

````python
# ## Python Memory Reference (Simple Explanation)

# In Python, every variable **does not store the value directly**.
# Instead, it stores a **reference (address)** to the value in memory.

# ---

# ## 1. What is Memory Reference?

# A **memory reference** means:

# > The location in memory where the object is stored.

# ---

# ## 2. Example

# ```python id="m1"
# a = 10
# b = 10

# print(id(a))
# print(id(b))
# ```

# 👉 Output will show same or similar memory address

# ---

# ## 3. Why same memory?

# Python **optimizes memory** for small integers and strings.
# So `a` and `b` point to the **same object in memory**.

# ---

# ## 4. Example with lists (different memory)

# ```python id="m2"
# a = [1, 2, 3]
# b = [1, 2, 3]

# print(id(a))
# print(id(b))
# ```

# 👉 Different memory addresses because lists are mutable objects.

# ---

# ## 5. Reference concept

# ```python id="m3"
# a = 10
# b = a

# print(id(a))
# print(id(b))
# ```

# 👉 Both refer to same memory location

# ---

# ## 6. Important Concept: Mutable vs Immutable

# ### Immutable (same memory reused sometimes)

# * int
# * string
# * tuple

# ### Mutable (new memory always)

# * list
# * dict
# * set

# ---

# ## 7. Real Meaning

# * Variable = label
# * Memory = actual storage
# * Reference = pointer to memory

# ---

# ## Interview Answer

# > "Python memory reference means variables do not store values directly, they store references to objects in memory. We can check this using the `id()` function. Immutable objects may share memory, while mutable objects usually have different memory locations."

# ---

# ## Simple Trick

# * Value → stored in memory
# * Variable → points to memory
# * `id()` → shows memory reference

# ---

````
